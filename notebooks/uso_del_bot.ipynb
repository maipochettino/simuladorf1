# 📊 Uso del bot de F1

Leemos `data/historial.csv`, el archivo donde el bot guarda cada interacción
(timestamp, chat_id, nombre, tipo, mensaje), y calculamos algunas métricas de uso.

```python
import os
import pandas as pd

# El notebook está en notebooks/, por eso subimos una carpeta
RUTA = os.path.join("..", "data", "historial.csv")

if not os.path.exists(RUTA):
    print("⚠️ Todavía no existe", RUTA, "- usá el bot un rato y volvé a correr el notebook.")
else:
    df = pd.read_csv(RUTA, encoding="utf-8")
    df["timestamp"] = pd.to_datetime(df["timestamp"])
    display(df.tail())
```

## Totales

```python
print("Total de mensajes:", len(df))
print("Usuarios únicos:", df["chat_id"].nunique())
```

## Comandos más usados

Nos quedamos con la primera palabra del mensaje (ej: `/estrategia 2024 belgica HAM` → `/estrategia`).
En grupos Telegram agrega el nombre del bot (`/historial@mi_bot`), así que lo sacamos.

```python
comandos = df[df["tipo"] == "comando"]["mensaje"]
comandos = comandos.str.split().str[0].str.split("@").str[0].str.lower()
comandos.value_counts().rename("veces").to_frame()
```

## Mensajes por día

```python
df["timestamp"].dt.date.value_counts().sort_index().rename("mensajes").to_frame()
```

## Mensajes por tipo

```python
df["tipo"].value_counts().rename("mensajes").to_frame()
```
