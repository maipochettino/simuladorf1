```python
position_data.info()
print("Registros:", len(position_data))
print("Pilotos:", position_data["DriverNo"].nunique())
```

```python
timing = session.get_data(dataNames="TimingData")
print(type(timing))
print(timing.head())
```

```python
tyres = session.get_data(dataNames="TyreStintSeries")
print(tyres.head())
```

```python
print(timing[
    ["DriverNo",
     "timestamp",
     "NumberOfLaps",
     "Position",
     "LastLapTime_Value",
     "GapToLeader",
     "IntervalToPositionAhead_Value"]
].head(20))
print(timing["NumberOfLaps"].unique()[:30])
```

```python
timing_laps = timing[
    timing["NumberOfLaps"].notna()
]

print(timing_laps[
    ["DriverNo",
     "timestamp",
     "NumberOfLaps",
     "Position",
     "LastLapTime_Value",
     "GapToLeader",
     "IntervalToPositionAhead_Value"]
].head(30))

print(timing_laps["NumberOfLaps"].value_counts().sort_index().head(20))
```

```python
print("Pilotos:", timing_laps["DriverNo"].nunique())
print("Vuelta máxima:", timing_laps["NumberOfLaps"].max())
print("Registros:", len(timing_laps))

vueltas = timing_laps[
    [
        "DriverNo",
        "timestamp",
        "NumberOfLaps",
        "Position",
        "LastLapTime_Value",
        "GapToLeader",
        "IntervalToPositionAhead_Value"
    ]
].copy()

vueltas = vueltas.rename(columns={
    "DriverNo": "piloto",
    "timestamp": "timestamp",
    "NumberOfLaps": "vuelta",
    "Position": "posicion",
    "LastLapTime_Value": "tiempo_vuelta",
    "GapToLeader": "gap_lider",
    "IntervalToPositionAhead_Value": "gap_adelante"
})

print(vueltas.head())
```

```python
vueltas = vueltas[
    vueltas["tiempo_vuelta"].notna()
].copy()

print(vueltas.head())
print("Registros:", len(vueltas))
```

```python
print(
    vueltas[vueltas["vuelta"] == 2][
        ["piloto", "vuelta", "posicion", "tiempo_vuelta",
         "gap_lider", "gap_adelante"]
    ]
)

print(
    vueltas.groupby("vuelta")["piloto"].nunique().head(15)
)
```

```python
print(tyres.columns.tolist())
print(tyres.info())
print(
    tyres[
        ["DriverNo", "timestamp", "PitCount", "Compound",
         "New", "TyresNotChanged", "TotalLaps", "StartLaps"]
    ].head(30)
)
```

```python
tyres_stints = tyres[
    tyres["Compound"].notna()
].copy()

print(tyres_stints[
    ["DriverNo", "timestamp", "PitCount", "Compound",
     "New", "TyresNotChanged", "TotalLaps", "StartLaps"]
].to_string(index=False))

print(
    tyres_stints.groupby("DriverNo")["Compound"].count()
)

print(
    tyres_stints["Compound"].value_counts()
)
```
