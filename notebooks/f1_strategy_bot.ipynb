```python
tyres_known = tyres_stints[
    tyres_stints["Compound"] != "UNKNOWN"
].copy()

tyres_known = tyres_known.rename(columns={
    "DriverNo": "piloto"
})

tablas = []

for piloto in vueltas["piloto"].unique():

    vueltas_piloto = vueltas[
        vueltas["piloto"] == piloto
    ].sort_values("timestamp").copy()

    tyres_piloto = tyres_known[
        tyres_known["piloto"] == piloto
    ].sort_values("timestamp").copy()

    vueltas_piloto = pd.merge_asof(
        vueltas_piloto,
        tyres_piloto[
            ["timestamp", "PitCount", "Compound"]
        ],
        on="timestamp",
        direction="backward"
    )

    vueltas_piloto["piloto"] = piloto

    tablas.append(vueltas_piloto)

vueltas_con_neumaticos = pd.concat(
    tablas,
    ignore_index=True
)

vueltas_con_neumaticos = vueltas_con_neumaticos.sort_values(
    ["piloto", "vuelta"]
).reset_index(drop=True)

print(vueltas_con_neumaticos.head(20))
```

```python
def tiempo_a_segundos(t):
    if pd.isna(t):
        return pd.NA
    
    minutos, segundos = str(t).split(":")
    return int(minutos) * 60 + float(segundos)


vueltas_con_neumaticos["tiempo_seg"] = (
    vueltas_con_neumaticos["tiempo_vuelta"]
    .apply(tiempo_a_segundos)
)

vueltas_con_neumaticos[
    ["piloto", "vuelta", "tiempo_vuelta", "tiempo_seg", "PitCount", "Compound"]
].head(20)
```

```python
vueltas_con_neumaticos = vueltas_con_neumaticos.sort_values(
    ["piloto", "PitCount", "vuelta"]
).copy()

vueltas_con_neumaticos["vueltas_neumatico"] = (
    vueltas_con_neumaticos
    .groupby(["piloto", "PitCount"])
    .cumcount() + 1
)

vueltas_con_neumaticos[
    [
        "piloto",
        "vuelta",
        "Compound",
        "PitCount",
        "vueltas_neumatico",
        "tiempo_seg"
    ]
].head(20)
```

```python
vueltas_con_neumaticos["PitCount"] = pd.to_numeric(
    vueltas_con_neumaticos["PitCount"],
    errors="coerce"
)

vueltas_con_neumaticos["es_pit"] = (
    vueltas_con_neumaticos
    .groupby("piloto")["PitCount"]
    .diff()
    .fillna(0)
    .gt(0)
)

print(vueltas_con_neumaticos["es_pit"].value_counts())
```

```python
costo_pit_aprox = vueltas_pit["perdida_pit_aprox"].median()

print("Costo aproximado del pit stop:", round(costo_pit_aprox, 2), "segundos")
```

```python
COSTO_PIT = costo_pit_aprox

def simular_estrategia(total_vueltas, estrategia):
    """
    Simula una estrategia de carrera.

    estrategia:
        lista de tuplas (compuesto, cantidad_de_vueltas)
    """

    tiempo_total = 0
    paradas = 0

    for compuesto, cantidad_vueltas in estrategia:

        for edad_neumatico in range(1, cantidad_vueltas + 1):
            tiempo_total += calcular_tiempo_vuelta(
                compuesto,
                edad_neumatico
            )

        paradas += 1

    # No hay parada después del último stint
    paradas -= 1

    tiempo_total += paradas * COSTO_PIT

    return tiempo_total
```

```python
resultados_estrategias = []

for vuelta_pit in range(10, 36):

    vueltas_medium = vuelta_pit
    vueltas_hard = 44 - vuelta_pit

    estrategia = [
        ("MEDIUM", vueltas_medium),
        ("HARD", vueltas_hard)
    ]

    tiempo_total = simular_estrategia(
        44,
        estrategia
    )

    resultados_estrategias.append({
        "vuelta_pit": vuelta_pit,
        "vueltas_medium": vueltas_medium,
        "vueltas_hard": vueltas_hard,
        "tiempo_total": tiempo_total
    })

resultados_estrategias = pd.DataFrame(resultados_estrategias)

resultados_estrategias.head()
```
