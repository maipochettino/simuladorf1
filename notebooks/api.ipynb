```python
import requests
import pandas as pd
```

```python
url = "https://api.openf1.org/v1/sessions"

params = {
    "year": 2024,
    "country_name": "Belgium",
    "session_name": "Race"
}

response = requests.get(url, params=params)

print(response.status_code)
print(response.json())
```

```python
url = "https://api.openf1.org/v1/laps"

params = {
    "session_key": 9574
}

response = requests.get(url, params=params)

print(response.status_code)

laps_openf1 = pd.DataFrame(response.json())

laps_openf1.head()
```

```python
laps_16_openf1 = laps_openf1[
    laps_openf1["driver_number"] == 16
][
    [
        "driver_number",
        "lap_number",
        "lap_duration",
        "is_pit_out_lap"
    ]
].copy()

laps_16_openf1.head(15)
```

```python
url = "https://api.jolpi.ca/ergast/f1/2024/12/results/"

response = requests.get(url)

print(response.status_code)

datos_jolpica = response.json()

datos_jolpica
```

```python
resultados = datos_jolpica["MRData"]["RaceTable"]["Races"][0]["Results"]

resultados_jolpica = pd.DataFrame(resultados)

resultados_jolpica.head()
```

```python
resultados_jolpica_simple = resultados_jolpica[
    ["number", "position", "points", "grid", "laps", "status"]
].copy()

resultados_jolpica_simple
```
